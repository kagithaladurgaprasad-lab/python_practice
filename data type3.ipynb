## list operators /methods

```python
#repeation 
a=[1,3,2,4,5]
print(a*2)
```

```python
#concatination
a=[1,3,2,4,5]
b=[5,6,7]
print(a+b)
```

```python
#append()
a=[1,2,3]
a.append(4)
print(a)
```

```python
# extend()---> appends the element of another list to the currrent list
a=[1,2,3]
b=[4,5,6]
a.extend(b)
print(a)
```

```python
a=[1,2,3]
b=[4,5,6]
b.extend(a)
print(b)
```

```python
#insert--> inserts an element an specified index
a=[1,2,3]
a.insert(1,1.5)   #(index,new value)
print(a)
```

```python
# Remove --> removes the first index in the specified index
a=[1,2,3,4,2,2]
a.remove(2)
print(a)
```

```python
#index---> returns the index of the first occurance of a specified element
a=[1,2,3]
b=a.index(3)
print("index position of 3:",b)
```

```python
#count--. returns the number of occurance of a specified in the list
a=[1,2,3,3,4,5]
b=a.count(3)
print(b)
```

```python
#sort()--> sorts the list in assending order
a=[1,3,9,2,4]
a.sort()
print(a)
```

```python
#reverse()---> Reverse the order of elemrnts in the list
a=[1,3,9,2,4]
a.sort()
a.reverse()
print(a)
```

```python
#copy()---> it will creates the copy of list
a=[1,2,3,4]
b=a.copy()
print(b)
```

```python
#clear()--> removes the all elements from the list
a=[1,2,3,4]
a.clear()
print(a)
```

```python
#min---> returns the smallest elements in the list
#max---> returns the largest
a=[12,34,56]
b=min(a)
print("min :",b)
c=max(a)
print("max :",c)
```

```python
#pop---> it removes and return the element in the specified index
a=[1,2,3,4]
b=a.pop(1)  #index position
print(a)
print(b)
```

```python
a=[1,2,3,4] 

#remove()
a.remove(4)
print(a)

#remove by index position 
a.pop(0)
print(a)

#delete by index
del a[1]
print(a)

#clear all elements
a.clear()
print(a)

#delete entire list
del a 
print(a)
```

```python
a=(1,1.5,1+3j,"hello")
print(a)
print(type(a))
```

```python
a=(1,)
print(type(a))
```

```python
# indexing
a=(1,1.5,1+3j,"hello")
print(a[0])
print(a[1])
print(a[2])
print(a[3])
#negative indexing
print(a[-1])
print(a[-2])
print(a[-3])
print(a[-4])
```

```python
#slicing
a=(10,20,30,40)
print(a[0:4])
print(a[0:7:2])
print(a[2:])
print(a[-6:-2])
print(a[::-1])
```

```python
#deleting a tuple
a=(1,2,3,4,5)
del a
```

```python
# remove an ele from tuple by convert into list
a=(1,2,3,4,5)
b=list(a)
b.remove(3)
c=tuple(b)
print(c)
```
