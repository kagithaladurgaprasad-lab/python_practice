```python
def is_prime(n):
    if n<=1:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return True
    return True
print(is_prime(5))
```

```python
import prime 
num=int(input("enter a number"))
if prime.is_prime(num):
    print("is a prime number")
else:
    print(num,"is not a prime number")
```

## Random module

```python
#random()--->it gives float number that ranges from 0.0 to 1.0
import random
num=random.random()
print(num)
```

```python
#randint()---> it genetare integer value from the range of numbers given
num=random.randint(1,500)
print(num)
```

```python
#randrange()--->selects an item or value from the given range(start:stop:stepsize)
num=random.randrange(1,10)
print(num)
```

```python
#random.choice()-->selects or make choice an item from a series at random
s=random.choice("prasad")
print(s)
s1=random.choice([1,2,3,4,5])
print(s1)
```

```python
#random.shuffle --->this function shuffile randomly 
a=[1,2,3,4]
random.shuffle(a)
print(a)
```

## stastics Module

```python
#-->find measures like mean ,medain ,mode,variance,standard devation
import statistics as st
data=[1,2,2,3,4,4,4,5]
mean_value=st.mean(data)
print(mean_value)

#Median:make sure given numbers must be in accending order
#it is used to return the middle value of a given data
#takes middle value
median=st.median(data)
print(median)

#mode :most frequent occurent value
most=st.mode(data)
print(most)


#variance: sum of squared difference/(n-1)
vari=st.variance(data)
print(vari)


#standard devation:square root of variance
sta=st.stdev(data)
print(sta)
```
