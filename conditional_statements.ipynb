## problem: encrypt a string by shifting characters

problem statement

write a python program that takes a string and an integer shift value as input ,then encrypts the string by shifting each character forward by shift character 

```python
for i in range(128):
    print(f"{i}--->{chr(i)}")
    
```

```python
text=input("enter the string:")
shift=int(input("enter shift value"))
encrypted=""
for char in text:
    if char.isalpha():
        print("it is an alphabate")
        if char.islower():
            base=ord('a')
        else:
            base=ord("A")
        shifted=(ord(char)-base+shift)%26+base
        encrypted_char= chr(shifted)
        print(f"Encrypted '{char}'--->'{encrypted_char}'")
        encrypted+=encrypted_char
    else:
        print(f"non alphabate character'{char}'remains unchanged.")
        encrypted+=char
print("\n Final Encrypted text:",encrypted)
```

```python
#palindrome---->that reads the same backword as forward
num=int(input("enter a number:"))
original=num
reverse=0
while num>0:
    digit=num%10
    reverse=reverse*10+digit
    num=num//10
if original==reverse:
    print("palindrome")
else:
    print("not a palindrome")

```

```python
s=input("enter a string")
reverse=""
length=len(s)-1
while length>=0:
    reverse=reverse+s[length]
    length-=1
if s==reverse:
    print("palindrome")
```

```python
## fibonnaci series using while loop

n=int(input("enter number of terms:"))
a,b=0,1
i=1
print("fibonnaci series:")
while i<=n:
    print(f"term{i}:{a}")
    c=a+b
    a=b
    b=c
    i+=1
```
