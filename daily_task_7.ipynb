```python
#7. List Reverser
#Problem Statement:
 #Write a program that reverses a given list using a for loop.

n=int(input("enter the number of elements:"))
li=[]
rev=[]
for i in range(n):
    a=input("enter the elements")
    li.append(a)
for i in range(-1,-(len(li)+1),-1):
    rev.append(li[i])
print(rev)
    
```

```python
#10. Sum of Odd Numbers up to N
#Problem Statement:
 #Write a program to calculate the sum of all odd numbers from 1 up to a given number N, using a for loop.
a=int(input("enter the no.of numbers"))
sum=0
for i in range(a+1):
    if i%2==0:
        continue
    else:
        sum=sum+i
print(sum)
    
```

```python
#11. Perfect Square Checker
#Problem Statement:
 #Write a program to check whether a given number is a perfect square, using a for loop.
a=int(input("enter a number"))
is_square=False
for i in range (1,a+1):
    if i*i==a:
       is_square=True
if is_square==True:
   print("perfect square")
else:    
    print("not a perfect suare")   
```

```python
#12. Prime Numbers up to N
#Problem Statement:
 #Write a program that prints all prime numbers up to a given number N, using a for loop.
n=int(input("enter the no.of numbers:"))
for n in range(2,n+1):
    is_prime=True
    for i in range(2,n):
        if n%i==0:
            is_prime=False
            break
    if is_prime:
        print(n,end=" ")
```

```python
#13. Fibonacci Series Generator
#Problem Statement:
 #Write a program that generates the Fibonacci series up to N terms, using a for loop.
n=int(input("enter the fibonacci number"))
a=0
b=1
print(a,b,end=" ")
for i in range(2,n):
    c=a+b
    a=b
    b=c
    print(c,end=" ")
    
```

```python
#14. Armstrong Number Checker
#Problem Statement:
 #Write a program to check whether a given number is an Armstrong number, using a for loop.
a=(input("enter a number"))
result=0
for ch in a:
    char=int(ch)
    b=char**3
    result=b+result
if result==int(a):
    print("amstrong")
else:
    print("not an amstrong")
    
```

```python
#15. Frequency Counter
#Problem Statement:
 #Write a program that counts how many times a given value appears in a list, using a for loop.
num=int(input("enter no.of elements"))
li=[]
count=0
for i in range(num):
    n=int(input("enter the element"))
    li.append(n)
fi=int(input("enter the element you want to find"))
for j in li:
    if li[j]==fi:
        count+=1
print(count)
        
```

```python
## 17. Common Elements Finder
#Problem Statement:
# Write a program that finds and prints common elements between two lists using a for loop.
num=int(input("enter no.of elements"))
li=[]
for i in range(num):
    n=int(input("enter the value"))
    li.append(n)
num1=int(input("enter second list no.of elements"))
lis=[]
for i in range(num1):
    n=int(input("enter the element"))
    lis.append(n)
res=[]
for j in li:
    if j in lis:
        res.append(j)
print(res)



    
```

```python
#19. Palindrome Word Checker
#Problem Statement:
 #Write a program to check whether a given word is a palindrome, using a for loop.
a=input("enter the a word: ")
reverse=""
for ch in a:
    reverse=reverse+ch
if reverse==a:
    print("palindrome")
else:
    print("not a palindrome")
    
```
