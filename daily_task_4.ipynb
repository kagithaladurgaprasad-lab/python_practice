```python
# Take a user’s age and classify the person as:
#Child (0–12)
#Teenager (13–19)
#Adult (20–60)
#Senior Citizen (above 60)
age=int(input("enter your age "))
if age<=12:
    print(" you are a child")
elif age<=19:
    print("you are a teenager")
elif age <=60:
    print("you are a adult")
else:
    print("you are a senior citizen")
```
