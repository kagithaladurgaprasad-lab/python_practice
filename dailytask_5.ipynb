```python
#1.  Scenario: Email Format Validator
#Create a program to check if the entered email is valid:
#Must contain @
#Must end with .com
#No starting blank spaces

a=input("enter your mail Id:")
if a.startswith(" "):
    print("invalid email Id")
elif "@" in a and a.endswith(".com"):
    print("valid email Id")

else:
    print("invalid email Id")

```

```python
#3.  Scenario: File Format Identifier
#Write a program to classify file type based on its extension:
a=input("enter the file type")
if a.endswith(".pdf"):
    print("PDF file")
elif a.endswith(".docx"):
    print("Word File")
elif a.endswith(".xlsx"):
    print("Excel File")
else:
    print("unknown format")

```

```python
#10.  Scenario: Character Category Checker
#Write a program to categorize a single character:
#Alphabet
#Digit
#Space
#Special Character

a=input("enter the charachter")
if a.isalpha():
    print("Alphabate")
elif a.isdigit():
    print("Number")
elif a==(" "):
    print("White spaces")
else:
    print("Spectial characters")

```

```python
#11.  Scenario: Country Code Identifier
#Identify the country based on phone number prefix:
#+91 ➞ India
#+44 ➞ UK
#Others ➞ Other
a=input("enter the code")
if len(a) and a.startswith("+91")and a[3:].isdigit():
    print("India")
elif len(a)and  a.startswith("+44") and a[3:].isdigit():
    print("UK")
else:
    print("other")



```

```python
#14.  Scenario: Word Length Categorizer
#Categorize a word based on its character count:
#Length ≤ 3 → Short
#Length 4–6 → Medium
#Length > 6 → Long
a=input("enter the word")
if len(a)<=3:
    print("Short")
elif len(a)<=6:
    print("Medium")
else:
    print("long")
```
