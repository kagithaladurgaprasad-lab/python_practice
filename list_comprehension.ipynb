```python
#without list comprehension
squares=[]
for i in range (1,6):
    squares.append(i**2)
print(squares)
```

```python
#with list comprehension
squares=[i**2 for i in range(1,6)]
print(squares)
```

```python
#generate a list with even numbers
even=[]
for i in range(1,11):
    if i%2==0:

       even.append(i)
print(even)
```

```python
#using list comprehension
even=[i for i in range(1,11) if i%2==0]
print(even)
```

```python
#without list compe covert all strings to the upper case
names=['apple','mango','banana']
new=[i.upper() for i in names]
print(new)
```

```python
pairs=[(i,i**2)for i in range (1,6)]
print(pairs)
```

```python
#Nested list comprehension
matrix=[[i*j for j in range(1,4)] for i in range(1,4)]
print(matrix)
```

```python
#create a list of all numbers from 1 to 50 that are divisible by both 3 and 5)
divi=[i for i in range (1,51) if i%3==0 and i%5==0]
print(divi)
```

## dictionary comprehension

```python
#create a dict of squares
squares={i:i**2 for i in range(1,6)}
print(squares)
```

```python
#convert list to dictionary with length
#{names:length}
names=['apple','mango','banana']
name_length={name:len(name) for name in names}
print(name_length)
```

```python
#filter dicit to include only even values
original={1:2,2:3,5:6,7:9}
even_values={k:v for k,v in original.items() if v%2==0}
print(even_values)
```

```python
#swap keys and values
original={'a':1,'b':2,'c':3}
swaped={v:k for k,v in  original.items()}
print(swaped)
```

```python
## given a string create a dictionary where ecah character is the key 
#and its value is the no.of times appear 
text="hello world"
count_letters={ch:text.count(ch) for ch in text if ch.isalpha()}
print(count_letters)
```
