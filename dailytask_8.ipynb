```python
#Remove prefix and suffix from string
 #Description: Remove specified prefix and suffix, e.g., '##' from '##Data-Science##'.
 #Input: '##Data-Science##'
 #Output: 'Data-Science'
a=input("enter the text")
result=""
for ch in a:
    if ch.isalpha():
        result+=ch
    elif ch=="-":
        result+="-"
print(result)
```
