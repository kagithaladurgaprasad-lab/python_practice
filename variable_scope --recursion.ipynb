## local variable/ scope


```python
def greet():
    message="hello"# local variable---> it can access only within the function
    print(message)
greet()
#print(message)--->it won't work because message is a local variable
```

```python
def show_name():
    name='jhon'
    print("Name is :",name)
show_name()
```

```python
def calculate_area(radius):
    pi=3.14159
    area=pi*radius**2
    return area
calculate_area(6)
```

## GLOBAL VARIABLE

```python
app_name='flipkart'# global variable--> we can access inside and outside the function
def show_name():
    print('application',app_name)
show_name()
print('app:',app_name)
```

```python
app_name='flipkart'# global variable--> we can access inside and outside the function
def show_name():
    global app_name  #by using global keyword we can modify variable_name
    app_name='Amazon'
    print("Application:",app_name)
   
show_name()
print('app:',app_name)
```

```python
status='offline'
def go_online():
    global status
    status='Online'
    print("status:",status)
go_online()
print("Status2:",status)
```

## ENCLOSING VARIABLE

```python
def outer():
    x=10  #this is local to 'oter()'and enclosing(non-local) for 'inner()'
    def inner():
        print(x)
    inner()
outer()
```

## LEBG rule

```python
pi=3.14159 # global variable

def geometry():
    pi=3.14 # enclosing variable
    def area(radius):
        #local name space
        shape='circle' # local variable
        result=pi*radius**2
        print(f"pi value:{pi}")
        print(f"shape:{shape}")
        print(f"Area of circle:{result}")
    area(5)
geometry()
```

## RECURSION

```python
#WITHOUT RECCURSION
def factorial_iter(n):
    result=1
    for i in range (1,n+1):
        result*=i
    return result
factorial_iter(7)
```

```python
#with recursion

def factorial_recursive(n):
    if n==0:
        return 1
    else:
        return n*factorial_recursive(n-1)
factorial_recursive(5)
```

```python
def rev_str(s):
    return s[::-1]
print(rev_str('hello'))
```

```python
def reveerse_string(s):
    if len(s)<=1:
        return s
    else:
        return s[-1]+reveerse_string(s[:-1])
print(reveerse_string('hello'))
```
