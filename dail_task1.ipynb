## PART  STRING CREATION AND INDEXING

```python
#Create an empty string and print it.
s=""
print(s)
print(type(s))
```

```python
#Print the first character of the string "Data".
s='Data'
print(s[0])
```

```python
#Print the last character of the string "Science".
s="Science"
print(s[-1])

```

```python
#Extract the character at index position 3 from "Programming".
s="Programming"
print(s[3])
```

```python
#Extract the first four characters from "Machine
s="Machine"
print(s[0:4])
```

```python
#From the string "Learning", extract the substring "earn"
s="Learning"
print(s[1:5])
print(s[-7:-3])
```

```python
#From the string "Artificial", extract "Art"
s="Artificial"
print(s[0:3])
print(s[-10:-7])
```

```python
#Extract the last three characters from "Intelligence".
s="Intelligence"
print(s[-3:])

```

```python
#Extract every second character from the string "Technology".
s="Technology"
print(s[::2])
```

```python
#Reverse the string "String"
s="String"
print(s[::-1])
```

## PART B Case Conversion

```python
#Convert the string "deep learning" into uppercase.
s= "deep learning"
print(s.upper())
```

```python
#Convert the string "DATA SCIENCE" into lowercase.
s="DATA SCIENCE"
print(s.lower())
```

```python
#Convert Convert "neural networks" so that only the first character is capitalized
s= "neural networks"
print(s.capitalize())
```

```python
#Convert "natural language processing" so that the first letter of every word is capitalized.
s="natural language processing"
print(s.title())
```

```python
#Change uppercase letters to lowercase and lowercase letters to uppercase in "Your_Name".
s="Your_Name"
print(s.swapcase())

```

```python
#Convert "SENSITIVE" into a case-insensitive version.
s="SENSITIVE"
print(s.lower())

```

## Part C: Search & Count Operations

```python
#Find the position of the character "i" in the string "intelligence"
s="intelligence"
print(s.index('i'))
print(s.find("i"))
```

```python
#Find the last position of the character "a" in the string "Canada"
s="Canada"
print(s.rfind('a'))
print(s.rindex('a'))
```

```python
#Count how many times the character "e" appears in "Deep Learning
s="Deep Learning"
print(s.count('e'))
```

```python
#Find the position of the character "n" in the string "Find"
s="Find"
print(s.find('n'))
```

```python
#Find the last position of the character "t" in the string "statistic"
s="statistic"
print(s.rfind("t"))
```

```python
#Count the number of spaces in the string "Machine Learning Model"
s="Machine Learning Model"
print(s.count(" " ))
```

## Part D: String Validation Checks 

```python
#Check whether the string "data123" contains only letters and numbers
s="data123" 
print(s.isalnum())
```

```python
#Check whether the string "OnlyLetters" contains only alphabetic characters.
s="OnlyLetters"
print(s.isalpha())
```

```python
#Check whether the string "123456" contains only numeric characters
s="123456"
print(s.isdigit())
```

```python
#Check whether the string " " contains only whitespace characters
s=" "
print(s.isspace())
```

```python
#Check whether the string "Name_" is written completely in uppercase
s="Name_" 
print(s.isupper())
```

```python
#Check whether the string "machine learning" is written completely in lowercase.
s="machine learning"
print(s.islower())
```

```python
#Check whether the string "Deep Learning" follows title-case format.
s="Deep Learning"
print(s.istitle())
```

```python
#Check whether the string "data_science" is a valid identifier
s= "data_science"
print(s.isidentifier())
```

 ## Part E: String Replacement 

```python
#Replace all occurrences of the character "a" with "@" in the string "Data Analysis"
s= "Data Analysis"
print(s.replace('a','@'))
```

```python
#Replace all spaces with underscores in the string "deep learning
s="deep learning"
print(s.replace(' ',"_"))
```

## Part F: Split & Combine Strings

```python
#Split the string ",Java,C++" into individual elements using the comma as a separator.
s=",Java,C++"
print(s.split(" ,"))
```

```python
#Convert the list ["I", "love", "Python"] into a single string as
list= ["I", "love", "Python"]
print(" ".join(list))

```
