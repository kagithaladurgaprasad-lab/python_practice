## Exception Handling

```python
 try:
     a=int(input("enter a number"))
     print(10/a)
 except ZeroDivisionError:     #if you the error
    print("cannot divide with zero")
 except Exception as e:  #if don't know the error
     print("error")
```

```python
try:
    a=int(input("enter a number"))
    print(10/a)
except ZeroDivisionError:     #if you the error
    print("cannot divide with zero")
except Exception as e:  #if don't know the error
    print("error")
else:
    print("Divison worked perfectly") #if code run success
finally:
    print("Program finished running") #wheather it takes error or run without error 
    
```

```python
user=input("Enter numbers seperated by space:").split()
numbers=[]
for i in user:
    numbers.append(int(i))
print(numbers)
```

```python
try:
    user=input("Enter numbers seperated by space:").split()
    numbers=[]
    for i in user:
        try:
            numbers.append(int(i))
        except ValueError:
           print(f"Invalid value:{i}")
    total=sum(numbers)
except Exception as e :
    print("Unexpected Error:",e)
else:
    print("sum of valid numbers:",total)
finally:
    print("processing completed")
```
