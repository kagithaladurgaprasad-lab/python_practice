## Static Method

```python
# Static Method
class student:
    
    school_name="ABC High School" #class variable

    def __init__(self,name,marks):
       self.name=name
       self.marks=marks
    @staticmethod  #it works like a normal function
    def calculate_marks(marks):
        if marks>=90:
            return "A"
        elif marks>=75:
            return "B"
        else:
            return "C"

s1=student("Jhon",90)
print(s1.calculate_marks(90))
s2=student("Ravi",85)
print(s2.calculate_marks(85))
```

## Access MOdifiers

```python
## public Access Modifiers
class student:
    def __init__(self,name,grade):
       self.name=name   #instance variable(public)
       self.grade=grade
    def display_info(self):   #instance method(public)
        print(f"StudentName:{self.name},Grade:{self.grade}")
s1=student("RAM","A")
print(s1.name,s1.grade)
s1.display_info()
```

```python
## private access modifier
class student:
    def __init__(self,name,grade):
       self.__name=name   #private variable
       self.__grade=grade #private variable
    def __calculate_result(self): #private method
        if self.__grade=='A':
            return "Excellent"
        elif self.__grade=="B":
            return "Good"
        else:
            return "Needs Improvement"
    def show_result(self):
        print(f"Student_name:{self.__name},Grade:{self.__grade}")
        print("Result:",self.__calculate_result())
        
s1=student("ravi","B")
s1.show_result()
```

```python
## protected access modifier
class student:
    def __init__(self,name,grade):
       self._name=name   #private variable
       self._grade=grade   #private variable

    def _display_info(self):
        print(f"StudentName:{self._name},Grade:{self._grade}")
s1=student("ram","A")
#print(s1._name)  ---> still works ,but should be avoided
#s1._display_info()
```

## INHERITANCE

```python
class animal:
    def eat(self):
        print("Eat")
    def sleep(self):
        print("Sleep")

#child class 
class dog(animal):# Dog inherits from animal
    def bark(self):
        print("barks")
class cat(animal):# cat inherits from animal
    def meow(self):
        print("meow")
Dog=dog()
Dog.eat()
Dog.sleep()
Dog.bark()


Cat=cat()

Cat.eat()
Cat.sleep()
Cat.meow()
```

## types of inheritance


## single inheritance

```python
class father:
    def drive(self):
        print("Driving")

class son(father):
    def study(self):
        print("study")
s=son()
s.drive()
s.study()
```

## multiple inheritance

```python
class father:
    def driving(self):
        print("Driving")
class mother:
    def cooking(self):
        print("cooking")
class child(father,mother):
    def study(self):
        print("study")
c=child()
c.driving()
c.cooking()
c.study()
```

## multilevel inheritance

```python
class grandfather:
    def farming(self):
        print("farming")
class father(grandfather):
    def driving(self):
        print("driving")
class son(father):
    def study(self):
        print("study")
s=son()
s.farming()
s.driving()
s.study()
```

## hierarchical inheritance

```python
class parent:
    def family_name(self):
        print("family name")
class son(parent):
    def play_cricket(self):
        print("playing cricket")
class daughter (parent):
    def dance(self):
        print("dancing")
s=son()
s.family_name()
s.play_cricket()
d=daughter()
d.family_name()
d.dance()
```

## hybrid inheritance

```python
class grandfather:
    def farming(self):
        print("farming")
class father(grandfather):
    def driving(self):
        print("driving")
class mother:
    def cooking(self):
        print("cooking")
class son (father,mother):
    def study(self):
        print("study")
s=son()
s.farming()
s.driving()
s.cooking()
s.study()
```

## method overriding

```python
class animal:
    def speak(self):
        print("some sound")
class dog(animal):
    def speak(self):
        print("bark")
d=dog()
d.speak()
```

```python
#Super()---->used inside achild class to call methods from its parent class
class animal:
    def speak(self):
        print("some sound")
class dog(animal):
    def speak(self):
        super().speak()  #call parent method
        print("bark")
d=dog()
d.speak()
```

## Inheritance--->Accessing Variables and Methods

```python
class parent:
    def __init__(self):
        self.name="parent class"
        self.value=100
    def show(self):
        print("This is parent class method")
        print("Name:",self.name)
        print("Value:",self.value)
class child(parent):
    def __init__(self):
        super().__init__()
        self.child_name="child class"
    def display(self):
        print("This is child method")
        print("Name:",self.child_name)

obj=child()
obj.show()
obj.display()
```

## Encapsulation 

```python
class bankaccount:
    def __init__(self,owner,balance):
        self.__owner =owner
        self.__balance=balance

    def get_owner(self):
        return self.__owner
    def get_balance(self):
        return self.__balance
    def deposite(self,amount):
        if amount>0:
            self.__balance +=amount
            print(f"Deposited {amount},New balance={self.__balance}")
        else:
            print("Deposited amount must be positive")
    def withdraw(self,amount):
        if 0<amount<=self.__balance:
            self.__balance-=amount
            print(f"withdraw {amount},Remaining balance={self.__balance}")
        else:
            print("Invalid withdrawl amount")
account=bankaccount("jhon",1000)
print(account.get_owner())
print(account.get_balance())

account.deposite(200)
print(account.get_balance())

account.withdraw(300)
print(account.get_balance())
```
